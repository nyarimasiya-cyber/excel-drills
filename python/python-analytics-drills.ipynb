## 3. Strings
*Python basics* — Joining, formatting, cleaning, slicing and splitting text.

| In Excel | In Python |
|---|---|
| `=A1&" "&B1` | `a + " " + b  or  f"{a} {b}"` |
| `=LEN(A1)` | `len(s)` |
| `=TRIM(A1)` | `s.strip()  (ends only)` |
| `=UPPER / LOWER / PROPER` | `s.upper() / s.lower() / s.title()` |
| `=SUBSTITUTE(A1,"R","")` | `s.replace("R", "")` |
| `=LEFT(A1,3)` | `s[:3]` |
| `=RIGHT(A1,4)` | `s[-4:]` |
| `=MID(A1,5,4)` | `s[4:8]` |
| `=TEXTSPLIT(A1,";")` | `s.split(";")` |
| `=TEXT(A1,"#,##0.00")` | `f"{a1:,.2f}"` |
| `=ISNUMBER(SEARCH("x",A1))` | `"x" in s  (case-sensitive)` |

### 3.1 Joining and length
Text goes in quotes. `+` joins strings and `len()` counts characters, spaces included.

```python
first = "Thandi"
last = "Nkosi"
full = first + " " + last
print(full)
print(len(full))
```

### 3.2 f-strings: formatting for reports
Put `f` before the quote and variables inside `{}`. After a colon you add a format: `,.2f` gives thousands separators and 2 decimals, `.1%` gives a percentage.

```python
client = "Acme Traders"
total = 12500.756
print(f"Invoice for {client}: R{total:,.2f}")
print(f"Growth: {0.0834:.1%}")
```

### 3.3 Cleaning text
Methods can be chained left to right. This is the Python version of nesting TRIM inside PROPER.

```python
raw = "  cape TOWN  "
print(raw.strip().title())
amount = "R1 250,00"
print(amount.replace("R", "").replace(" ", "").replace(",", "."))
```

### 3.4 Indexing and slicing
Positions start at 0. `s[a:b]` takes from position a up to, but not including, b. Negative numbers count from the end.

```python
code = "INV-2026-0042"
print(code[0])
print(code[:3])
print(code[4:8])
print(code[-4:])
```

### 3.5 Splitting and searching
`split()` breaks text into a list. `in` checks whether text contains something.

```python
line = "Cape Town;A4 Paper;20"
parts = line.split(";")
print(parts)
print(parts[1])
print("Paper" in line)
```

## 4. Lists and dictionaries
*Python basics* — Holding many values: lists for columns, dictionaries for lookups, sets for unique values.

| In Excel | In Python |
|---|---|
| `A column of values` | `a list: [42000, 38500, ...]` |
| `=COUNTA(A:A)` | `len(lst)` |
| `=SUM / MAX / MIN` | `sum(lst) / max(lst) / min(lst)` |
| `=SORT(A2:A9,,-1)` | `sorted(lst, reverse=True)` |
| `=INDEX(A2:A9,1)` | `lst[0]` |
| `XLOOKUP on a 2-column table` | `d[key]` |
| `=IFERROR(XLOOKUP(...),0)` | `d.get(key, 0)` |
| `=UNIQUE(A2:A9)` | `sorted(set(lst))` |

### 4.1 Lists
A list holds values in order inside square brackets. It behaves much like a single Excel column.

```python
monthly = [42000, 38500, 51200, 47800]
monthly.append(53900)
print(monthly)
print(monthly[0], monthly[-1])
print(len(monthly), sum(monthly), max(monthly))
print(sorted(monthly, reverse=True))
print(monthly[1:3])
```

### 4.2 Dictionaries: built-in lookups
A dictionary maps keys to values, like a two-column lookup table. `d[key]` fails if the key is missing; `d.get(key, default)` returns a fallback instead.

```python
prices = {"A4 Paper": 89.0, "Pens": 4.5, "Stapler": 145.0}
print(prices["Pens"])
prices["Notebooks"] = 22.0
print(prices.get("Tape", 0))
print(list(prices.keys()))
```

### 4.3 A record is a dictionary
One row of data fits naturally in a dictionary. A list of these dictionaries is a table, which is exactly what pandas will make easier.

```python
order = {"branch": "Durban", "product": "Pens", "qty": 90}
print(order["qty"] * prices[order["product"]])
```

### 4.4 Sets for unique values
A set keeps each value once and has no order. Wrap it in `sorted()` for a tidy list.

```python
branches = ["Durban", "Cape Town", "Durban", "Johannesburg", "Cape Town"]
print(sorted(set(branches)))
print(len(set(branches)))
```

## 5. Conditions
*Python basics* — Making decisions with if, elif and else.

| In Excel | In Python |
|---|---|
| `=IF(A1>=50000,"Hit","Miss")` | `"Hit" if a >= 50000 else "Miss"` |
| `Nested IF / IFS` | `if / elif / else` |
| `=AND(a,b) / =OR(a,b) / =NOT(a)` | `a and b / a or b / not a` |
| `A1=B1 (compare)` | `a == b` |
| `A1<>B1` | `a != b` |

### 5.1 Comparisons give True or False
`==` compares, a single `=` assigns. Text comparisons are case-sensitive.

```python
qty = 120
print(qty > 100)
print(qty == 100)
print("Pens" != "pens")
```

### 5.2 if / elif / else
Python checks each condition in order and runs the first block that is True. The colon and the 4-space indent are required: indentation is how Python knows what belongs to the if.

```python
revenue = 48000
if revenue >= 50000:
    status = "On target"
elif revenue >= 40000:
    status = "Close"
else:
    status = "Behind"
print(status)
```

### 5.3 Combining with and / or
`and` needs both sides True. `or` needs at least one.

```python
branch = "Durban"
qty = 150
if branch == "Durban" and qty > 100:
    print("Bulk Durban order")
if branch == "Cape Town" or qty > 100:
    print("Flag for review")
```

### 5.4 One-line IF
For simple two-way choices, the inline form reads like Excel's IF.

```python
discount = 0.1 if qty >= 100 else 0
print(discount)
```

## 6. Loops
*Python basics* — Repeating work with for and while, and the compact list comprehension.

| In Excel | In Python |
|---|---|
| `Drag a formula down a column` | `for item in lst:` |
| `Running-total helper column` | `total += value inside a loop` |
| `=COUNTIF(A:A,">=100")` | `loop with if and a counter` |

### 6.1 for loops
A for loop runs its indented block once for each item. The variable after `for` holds the current item.

```python
branches = ["Cape Town", "Johannesburg", "Durban"]
for b in branches:
    print(b.upper())
```

### 6.2 Accumulating a total
Start a variable at 0 before the loop and add to it inside.

```python
sales = [42000, 38500, 51200, 47800]
total = 0
for s in sales:
    total += s
print(total)
```

### 6.3 range()
`range(start, stop)` counts from start up to, but not including, stop.

```python
for month in range(1, 4):
    print(f"Month {month}")
```

### 6.4 Loop plus if
Combine a loop with a condition to filter or count.

```python
big = []
for s in sales:
    if s > 45000:
        big.append(s)
print(big)
print(len(big))
```

### 6.5 Looping over a dictionary
`.items()` gives you each key and value together.

```python
stock = {"A4 Paper": 12, "Pens": 340, "Stapler": 3}
for item, units in stock.items():
    if units < 10:
        print(f"Reorder {item}: only {units} left")
```

### 6.6 while loops
A while loop repeats until its condition turns False. Make sure something inside the loop moves it toward False.

```python
balance = 10000
months = 0
while balance > 0:
    balance -= 2500
    months += 1
print(months)
```

### 6.7 List comprehensions
A one-line loop that builds a new list. In pandas you will rarely loop at all, because whole columns are calculated at once.

```python
with_vat = [round(s * 1.15) for s in sales]
print(with_vat)
print([s for s in sales if s > 45000])
```

### 7.1 def and return
`def` creates a function. `return` sends a value back to whoever called it.

```python
def add_vat(amount):
    return amount * 1.15

print(add_vat(1000))
print(round(add_vat(89.9), 2))
```

### 7.2 Several inputs and defaults
Parameters with `=` have a default, so the caller can leave them out.

```python
def margin(revenue, cost, as_percent=True):
    m = (revenue - cost) / revenue
    return round(m * 100, 1) if as_percent else m

print(margin(5000, 3200))
print(margin(5000, 3200, as_percent=False))
```

### 7.3 Functions with rules
Put business rules in one function so every report uses the same logic. A `return` ends the function immediately.

```python
def size_band(qty):
    if qty >= 100:
        return "Bulk"
    elif qty >= 20:
        return "Medium"
    return "Small"

print([size_band(q) for q in [5, 35, 150]])
```

### 7.4 lambda: one-line functions
Handy for short calculations, especially inside pandas `.apply()`.

```python
commission = lambda sales: sales * 0.03
print(commission(40000))
```

### 7.5 Importing libraries

```python
import math
import statistics
print(math.ceil(7.2))
print(statistics.median([3, 9, 4, 7]))
```

## 8. NumPy arrays
*NumPy* — Fast whole-column maths, statistics, filters and vectorised IF.

| In Excel | In Python |
|---|---|
| `Spill: =B2:B5*C2:C5` | `qty * price` |
| `=AVERAGE / =MEDIAN` | `a.mean() / np.median(a)` |
| `=STDEV.P` | `a.std()  (pandas .std() is STDEV.S)` |
| `=FILTER(B2:B7,B2:B7>45000)` | `a[a > 45000]` |
| `=COUNTIF(B2:B7,">45000")` | `(a > 45000).sum()` |
| `=IF(B2:B7>=45000,"Hit","Miss")` | `np.where(a >= 45000, "Hit", "Miss")` |
| `Running total` | `a.cumsum()` |
| `=XMATCH(MAX(B:B),B:B)` | `a.argmax()  (0-based)` |
| `Column totals / row totals` | `m.sum(axis=0) / m.sum(axis=1)` |

### 8.1 Arrays: maths on whole columns
A NumPy array is like a list built for numbers. Arithmetic applies to every item at once, with no loop. This is the same idea as an Excel spill formula.

### 10.1 Missing values
pandas shows blanks as NaN. Count them with `isna().sum()`, then decide: fill with a value or drop the row. Work on a `.copy()` so the original stays untouched.

```python
print(sales["qty"].isna().sum())
clean = sales.copy()
clean["qty"] = clean["qty"].fillna(0)
print(clean["qty"].isna().sum())
```

### 10.2 Calculated columns
Assign to a new column name. The calculation runs on every row at once.

```python
clean["revenue"] = clean["qty"] * clean["unit_price"]
print(clean[["product", "qty", "unit_price", "revenue"]].head(4))
```

### 10.3 Dates
Dates imported as text can't be grouped by month. Convert with `pd.to_datetime`, then pull parts out with `.dt`.

```python
clean["date"] = pd.to_datetime(clean["date"])
clean["month"] = clean["date"].dt.month_name()
clean["weekday"] = clean["date"].dt.day_name()
print(clean[["date", "month", "weekday"]].head(3))
```

### 10.4 Messy text and numbers stored as text
Every string method from module 3 works on a whole column through `.str`. Here a rand amount with spaces and a comma decimal becomes a real number.

```python
messy = pd.DataFrame({
    "client": ["  acme traders", "ACME TRADERS ", "Bright Ink"],
    "amount": ["R1 250,00", "R980,50", "R2 100,00"],
})
messy["client"] = messy["client"].str.strip().str.title()
messy["amount"] = (messy["amount"]
                   .str.replace("R", "")
                   .str.replace(" ", "")
                   .str.replace(",", ".")
                   .astype(float))
print(messy)
print(messy["amount"].sum())
print(messy["client"].nunique())
```

## 13. Capstone: folder to report
*Automation* — The Power Query 'From Folder' workflow in Python: combine every file, clean, summarise, write a multi-sheet workbook.

| In Excel | In Python |
|---|---|
| `Get Data > From Folder` | `Path("folder").glob("*.xlsx")` |
| `Combine Files` | `pd.concat(list_of_frames)` |
| `Source.Name column` | `part["source_file"] = file.name` |
| `Applied Steps` | `a cleaning function` |
| `Close & Load to sheets` | `pd.ExcelWriter + to_excel(sheet_name=...)` |
| `Data > Refresh All` | `re-run the notebook or script` |

### 13.1 Step 1: a folder of monthly files
To keep the drill self-contained, this step splits `sales` into one Excel file per month. In real work these files already exist and you skip this step.

```python
from pathlib import Path

folder = Path("monthly_sales")
folder.mkdir(exist_ok=True)
month = pd.to_datetime(sales["date"]).dt.strftime("%Y-%m")
for m, part in sales.groupby(month):
    part.to_excel(folder / f"sales_{m}.xlsx", index=False)
print(sorted(p.name for p in folder.glob("*.xlsx")))
```

### 13.2 Step 2: read and combine every file
Loop over the folder, read each file, tag it with its file name, and stack them. A new month's file is picked up automatically on the next run.

```python
frames = []
for file in sorted(folder.glob("*.xlsx")):
    part = pd.read_excel(file)
    part["source_file"] = file.name
    frames.append(part)

combined = pd.concat(frames, ignore_index=True)
print(combined.shape)
print(combined["source_file"].value_counts().sort_index())
```

### 13.3 Step 3: cleaning as one function
Put every cleaning step in one function, the way Power Query keeps Applied Steps. Each new batch of data gets exactly the same treatment.

```python
def clean_sales(raw):
    out = raw.copy()
    out["qty"] = out["qty"].fillna(0)
    out["date"] = pd.to_datetime(out["date"])
    out["month"] = out["date"].dt.strftime("%Y-%m")
    out["revenue"] = out["qty"] * out["unit_price"]
    return out

data = clean_sales(combined)
print(data[["month", "branch", "revenue"]].head(3))
```

### 13.4 Step 4: summaries
Build every table the report needs.

```python
by_branch = data.groupby("branch").agg(orders=("order_id", "count"),
                                     revenue=("revenue", "sum")).round(2)
by_month = data.pivot_table(index="month", columns="branch", values="revenue",
                            aggfunc="sum", fill_value=0)
print(by_branch)
print(by_month)
```

### 13.5 Step 5: write a multi-sheet workbook
`ExcelWriter` writes several tables into one file, each on its own sheet. Download it from the Colab folder panel.

```python
with pd.ExcelWriter("branch_report.xlsx") as xw:
    by_branch.to_excel(xw, sheet_name="By branch")
    by_month.to_excel(xw, sheet_name="By month")
    data.to_excel(xw, sheet_name="Data", index=False)

print(pd.ExcelFile("branch_report.xlsx").sheet_names)
```
